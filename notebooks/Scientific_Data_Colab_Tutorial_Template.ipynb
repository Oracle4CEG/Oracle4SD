### Cell 2 — Reusable infrastructure

These helpers manage paths, checksums, logs and explicit errors. They contain no scientific
transformation. A checksum checks file identity, not whether a source measurement is correct.
The [Python hashlib documentation](https://docs.python.org/3/library/hashlib.html) explains the underlying hash API.

**Expected output:** “Helpers loaded.” If you restart the runtime, run Cells 1 and 2 again.


```python
import csv, hashlib, json, os, platform, re, subprocess, time, uuid, zipfile
from datetime import datetime, timezone
from urllib.parse import urlsplit
from urllib.request import Request, urlopen
from urllib.error import HTTPError, URLError
try:
    from IPython.display import display
except ImportError:
    def display(value):
        print(value)

class AuthorInputRequired(ValueError):
    pass

def require(value, label):
    if value is None or value == "" or value == [] or (
        isinstance(value, str) and "AUTHOR INPUT" in value
    ):
        raise AuthorInputRequired(f"AUTHOR INPUT required: {label}. See the preceding Markdown cell.")
    return value

def child(root, relative):
    root = Path(root).resolve()
    rel = Path(require(relative, "a relative file/folder path"))
    if rel.is_absolute():
        raise ValueError("Use a path relative to the documented root.")
    path = (root / rel).resolve()
    if path == root or not path.is_relative_to(root):
        raise ValueError("The requested path must remain inside its documented root.")
    return path

def sha256_file(path):
    h = hashlib.sha256()
    with Path(path).open("rb") as stream:
        for chunk in iter(lambda: stream.read(1024 * 1024), b""):
            h.update(chunk)
    return h.hexdigest()

def save_json(path, value):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    path.write_text(json.dumps(value, indent=2, ensure_ascii=False, allow_nan=False), encoding="utf-8")

def inventory(root):
    root = Path(root).resolve()
    rows = []
    for p in sorted(root.rglob("*")):
        if p.is_file():
            if p.is_symlink() or not p.resolve().is_relative_to(root):
                raise ValueError("Inventory encountered an unsupported linked file.")
            rows.append({"path": p.relative_to(root).as_posix(),
                         "bytes": p.stat().st_size, "sha256": sha256_file(p)})
    return rows

STATE = {"query_completed": False, "process_completed": False}
print("Helpers loaded.")
```

### Cell 5 — Connect code cells to the repository

Fill each required <code>STEPS</code> entry with the **real** source file and argument list.
Tokens substitute documented paths; the wrapper does not invent a command interface.
For API keys, use a secure interactive mechanism such as <code>getpass.getpass()</code>,
pass the value through an environment variable expected by your code, and remove it when done.
Do not print secrets or include them in command arguments, URLs, saved outputs or public logs.

This helper records execution, runtime and exit status. A completed command means the program
exited normally; scientific checks and output verification still follow.


```python
def run_repository_step(stage):
    entry = STEPS[stage]
    source = child(CODE_DIR, require(entry["source_path"], f"STEPS[{stage!r}].source_path"))
    if not source.is_file():
        raise FileNotFoundError(f"The source file for {stage} does not exist at the pinned commit.")
    argv = require(entry["argv"], f"STEPS[{stage!r}].argv")
    if not isinstance(argv, list) or not all(isinstance(x, str) for x in argv):
        raise TypeError("Use a list of command arguments, not a shell command string.")
    substitutions = {"python": sys.executable, "code_dir": str(CODE_DIR),
        "raw_dir": str(RAW_DIR), "processed_dir": str(PROCESSED_DIR),
        "reports_dir": str(REPORTS_DIR), "analysis_dir": str(ANALYSIS_DIR),
        "seed": str(CONFIG["seed"])}
    resolved = []
    for argument in argv:
        for key, value in substitutions.items():
            argument = argument.replace("{" + key + "}", value)
        resolved.append(argument)
    started = time.monotonic()
    # Review/sanitize your program's output before sharing the notebook.
    completed = subprocess.run(resolved, cwd=CODE_DIR, check=False)
    record = {"stage": stage, "source_path": entry["source_path"],
        "source_url": CONFIG["repo_url"] + "/blob/" + CONFIG["commit"] + "/" + entry["source_path"],
        "argv_without_secrets": resolved, "seconds": round(time.monotonic() - started, 3),
        "returncode": completed.returncode,
        "execution_status": "completed" if completed.returncode == 0 else "failed"}
    with (REPORTS_DIR / "execution_log.jsonl").open("a", encoding="utf-8") as stream:
        stream.write(json.dumps(record) + "\n")
    if completed.returncode:
        raise RuntimeError(f"{stage} failed. Inspect the error and execution log; do not use stale outputs.")
    print("Executed source:", record["source_url"])
    return record
```

```python
def fetch_archived(specs, root, byte_limit):
    rows, used, names = [], 0, set()
    for item in specs:
        for key in ["source_id", "url", "filename", "sha256", "citation_url"]:
            require(item.get(key), f"archived file {key}")
        url = item["url"]
        parsed = urlsplit(url)
        if parsed.scheme != "https" or parsed.username or parsed.password:
            raise ValueError("Use an authorized HTTPS file URL without embedded credentials.")
        if not re.fullmatch(r"[0-9a-fA-F]{64}", item["sha256"]):
            raise ValueError("Expected SHA-256 must contain 64 hexadecimal characters.")
        target = child(root, item["filename"])
        if target in names or target.exists():
            raise FileExistsError("Input filename already used. Start a fresh run or fix the manifest.")
        names.add(target)
        target.parent.mkdir(parents=True, exist_ok=True)
        partial = target.with_name(target.name + ".partial")
        if partial.exists():
            raise FileExistsError("An incomplete file already exists; inspect it before starting again.")
        for attempt in range(3):
            size = 0
            try:
                req = Request(url, headers={"User-Agent": "ScientificDataTutorial/1.0"})
                with urlopen(req, timeout=30) as response, partial.open("wb") as out:
                    if urlsplit(response.geturl()).scheme != "https":
                        raise ValueError("Download redirected away from HTTPS.")
                    for chunk in iter(lambda: response.read(1024 * 1024), b""):
                        size += len(chunk)
                        if used + size > byte_limit:
                            raise ValueError("Download exceeds the configured tutorial byte limit.")
                        out.write(chunk)
                if sha256_file(partial).lower() != item["sha256"].lower():
                    raise ValueError("Checksum mismatch. Verify the release URL and expected hash.")
                partial.rename(target)
                break
            except Exception as exc:
                partial.unlink(missing_ok=True)
                retryable = isinstance(exc, URLError) and (
                    not isinstance(exc, HTTPError) or exc.code in {429, 500, 502, 503, 504})
                if not retryable or attempt == 2:
                    raise
                time.sleep(2 ** attempt)
        used += size
        rows.append({"source_id": item["source_id"], "path": item["filename"],
            "citation_url": item["citation_url"], "download_url": url,
            "bytes": size, "sha256": sha256_file(target),
            "retrieved_utc": datetime.now(timezone.utc).isoformat(),
            "status": "checksum_matched"})
    return rows

STATE["query_completed"] = False
if CONFIG["mode"] == "archived":
    provenance = fetch_archived(ARCHIVED_FILES, RAW_DIR, CONFIG["max_download_bytes"])
else:
    if not CONFIG["allow_live"]:
        raise AuthorInputRequired("Live acquisition is disabled.")
    require(CONFIG["live_bounds_and_cost_note"], "live query limits and costs")
    run_repository_step("query")
    provenance = [{"status": "live_collection_executed",
                   "note": "Consult the project query log and source registry."}]
raw_inventory = inventory(RAW_DIR)
if not raw_inventory:
    raise RuntimeError("No raw files were obtained. Check the source and query logs.")
save_json(REPORTS_DIR / "acquisition_provenance.json", provenance)
save_json(REPORTS_DIR / "raw_inventory.json", raw_inventory)
STATE["query_completed"] = True
display(pd.DataFrame(raw_inventory))
print("Acquired files; inspect their meaning in Part IV.")
```

```python
def load_table(root, spec):
    require(spec, "a documented table specification or a modality-specific inspection adapter")
    path = child(root, require(spec.get("path"), "table path"))
    if not path.is_file():
        raise FileNotFoundError("The documented table file was not found.")
    if path.stat().st_size > CONFIG["max_table_bytes"]:
        raise ValueError("Use a smaller verified sample or the repository's streaming reader.")
    fmt = require(spec.get("format"), "table format")
    if spec.get("missing_values") is None or spec.get("dtypes") is None:
        raise AuthorInputRequired("Specify missing_values and dtypes, even if explicitly empty.")
    if fmt in {"csv", "tsv"}:
        df = pd.read_csv(path, sep=spec.get("sep", "\t" if fmt == "tsv" else ","),
                         encoding=spec.get("encoding", "utf-8"), dtype="string",
                         keep_default_na=False, na_values=spec["missing_values"])
    elif fmt in {"jsonl", "json_table"}:
        df = pd.read_json(path, lines=(fmt == "jsonl"), dtype=False, convert_dates=False)
        # For structured JSON with nested objects, use the repository's documented reader.
        if spec["missing_values"]:
            df = df.replace(spec["missing_values"], pd.NA)
    elif fmt == "parquet":
        df = pd.read_parquet(path)  # Requires the author's pinned Parquet engine.
        if spec["missing_values"]:
            df = df.replace(spec["missing_values"], pd.NA)
    else:
        raise AuthorInputRequired("Add the actual reader for this format; do not guess its schema.")
    for column, dtype in spec["dtypes"].items():
        if column not in df.columns:
            raise ValueError(f"Declared column is missing: {column}")
        df[column] = df[column].astype(dtype)
    return df

if not STATE["query_completed"]:
    raise RuntimeError("Complete Part III before inspecting raw data.")
display(pd.DataFrame(inventory(RAW_DIR)))
raw_df = None
if RAW_TABLE is not None:
    raw_df = load_table(RAW_DIR, RAW_TABLE)
    print("Loaded file:", RAW_TABLE["path"], "| rows:", len(raw_df), "| columns:", len(raw_df.columns))
    display(raw_df.head(5))
    display(pd.DataFrame({"field": raw_df.columns, "dtype": raw_df.dtypes.astype(str).values,
                          "missing": raw_df.isna().sum().values}))
else:
    run_repository_step("inspect_raw")
```

```python
if not STATE["process_completed"]:
    raise RuntimeError("Complete processing and inspection before this step.")
validation_rows = []

def add_count_check(check_id, n_tested, n_failed, unit, evidence, criterion):
    if n_tested is None or n_tested == 0:
        status, fraction = "not_evaluated", None
    else:
        if not isinstance(n_tested, int) or not isinstance(n_failed, int):
            raise TypeError("Count checks require integer denominators and failures.")
        if not 0 <= n_failed <= n_tested:
            raise ValueError("Failures must lie between zero and the number tested.")
        status = "pass" if n_failed == 0 else "fail"
        fraction = n_failed / n_tested
    row = {"check_id": check_id, "status": status, "n_tested": n_tested,
           "n_failed": n_failed, "unit": unit, "failure_fraction": fraction,
           "criterion": criterion, "evidence": evidence}
    validation_rows.append(row)
    return row

def validate_table(df, rules):
    if df is None:
        add_count_check("table_checks", None, None, "not_defined", None, "Requires an applicable table")
        return
    n = len(df)
    required = rules["required_columns"]
    if required is None:
        add_count_check("required_columns", None, None, "fields", None, "Not configured")
    else:
        absent = sorted(set(required) - set(df.columns))
        add_count_check("required_columns", len(required), len(absent), "declared fields",
                        {"missing_fields": absent}, "All declared fields exist")
    key = rules["unique_key"]
    if not key:
        add_count_check("unique_key", None, None, "rows", None, "Not configured")
    elif any(c not in df.columns for c in key):
        add_count_check("unique_key", None, None, "rows", {"key": key}, "Key columns absent")
    else:
        bad = df.duplicated(key, keep=False) | df[key].isna().any(axis=1)
        add_count_check("unique_key", n, int(bad.sum()), "rows",
                        {"columns": key, "bad_row_positions": np_positions(bad)[:20]},
                        "No missing key components or duplicate-key groups")
    non_null = rules["non_nullable"]
    if not non_null:
        add_count_check("non_nullable", None, None, "rows", None, "Not configured")
    else:
        for column in non_null:
            if column not in df.columns:
                add_count_check("non_null:" + column, None, None, "rows", None, "Field absent")
            else:
                mask = df[column].isna()
                add_count_check("non_null:" + column, n, int(mask.sum()), "rows",
                                {"bad_row_positions": np_positions(mask)[:20]}, "No missing values")
    expected = rules["expected_rows"]
    if expected is None:
        add_count_check("expected_row_count", None, None, "file counts", None, "Not configured")
    else:
        if not isinstance(expected, int) or expected < 0:
            raise ValueError("Expected rows must be a nonnegative integer from this release's evidence.")
        add_count_check("expected_row_count", 1, int(n != expected), "file counts",
                        {"expected": expected, "observed": n}, "Exact documented count for this scope")

def np_positions(mask):
    return [i for i, value in enumerate(mask.tolist()) if bool(value)]

validate_table(processed_df, TABLE_RULES)
display(pd.DataFrame(validation_rows))
print("These checks do not establish measurement accuracy or population representativeness.")
```

### Cell 17 — Compare with an earlier independent run

First finish one run and retain its processed manifest and outputs. Then have an independent
reader use a fresh session with the same declared inputs and code. Supply that earlier
manifest through <code>BASELINE_MANIFEST</code>.

This example tests **byte equality of processed files**, with missing and extra files included.
It refuses the current run as its own baseline. It cannot establish who performed a run.
For stochastic, floating-point or metadata-sensitive outputs, use justified numerical or
scientific comparisons in the repository and report their tolerance and uncertainty.
Do not silently replace a failed exact comparison with a more permissive criterion.


```python
def compare_manifests(current, baseline):
    for key in ["run_id", "code_commit", "data_version", "data_scope", "files"]:
        if key not in baseline:
            raise ValueError(f"Baseline manifest lacks {key}.")
    if baseline["run_id"] == current["run_id"]:
        raise ValueError("The baseline must come from a separate earlier run.")
    for key in ["code_commit", "data_version", "data_scope"]:
        if baseline[key] != current[key]:
            raise ValueError(f"Baseline and current run differ in {key}; comparison is not equivalent.")
    old = {x["path"]: x["sha256"] for x in baseline["files"]}
    new = {x["path"]: x["sha256"] for x in current["files"]}
    paths = sorted(set(old) | set(new))
    differences = [x for x in paths if old.get(x) != new.get(x)]
    return len(paths), len(differences), differences

if BASELINE_MANIFEST is None:
    add_count_check("separate_run_byte_comparison", None, None, "processed files",
                    None, "No earlier run manifest supplied")
else:
    baseline = json.loads(Path(BASELINE_MANIFEST).read_text(encoding="utf-8"))
    n, failed, differences = compare_manifests(processed_manifest, baseline)
    add_count_check("separate_run_byte_comparison", n, failed, "processed files",
                    {"different_or_missing_files": differences,
                     "baseline_run_id": baseline["run_id"]}, "Exact file-byte equality")
save_json(REPORTS_DIR / "validation_report.json",
          {"run_id": RUN_ID, "code_commit": CONFIG["commit"],
           "data_version": CONFIG["data_version"], "data_scope": CONFIG["data_scope"],
           "checks": validation_rows,
           "interpretation": "Individual check outcomes; no overall scientific-quality certification."})
pd.DataFrame(validation_rows).to_csv(REPORTS_DIR / "validation_summary.csv", index=False)
display(pd.DataFrame(validation_rows))
print("Report saved. Inspect all failures and all checks not evaluated.")
```
